## Series

<u>`Series`</u> is a one-dimensional labeled array capable of holding any data type (integers, string, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the **index**. The basic method to create a <u>`Series`</u> is to call:

<code> s = pd.Series(data, index=index)</code>

Here, `data` can be many different things:

* a Python dict
* an ndarray
* a scalar value (like 5)

The passed **index** is a list of axis labels. Thus, this separates into a few cases depending on what **data is**:

### From ndarray

if `data` is an ndarray, **index** mus be the same length as **data**. if no index is passed, one will be created having values `[0, ..., len(data) - 1]`.

```python
s = pd.Series(np.random.randn(5), index=["a", "b", "c", "d", "e"])
```

### From dict

<u>`Series`</u> can be instantiated from dicts:

```python
d = {"b": 1, "a": 0, "c": 2}
```

```python
pd.Series(d)
```

```python
pd.Series(d, index=["b", "c", "d", "a"])
```

### From scalar value

If `data` is a scalat value, an index must be provided. The value will be repeated to match the length of **index**.

```python
pd.Series(5.0, index=["a", "b", "c", "d", "e"])
```
