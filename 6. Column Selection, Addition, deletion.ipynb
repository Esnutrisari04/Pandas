## Column selection, addition, deletion

You can threat <u>`DataFrame`</u> semantically like a dict of like-indexed <u>`Series`</u> object. Getting, setting, and deleting columns works with the same syntax as the analogous dict operations:

```python
df = {
    "one": pd.Series([1, 2, 3], index=["a", "b", "c"]),
    "two": pd.Series([1, 2, 3, 4], index=["a", "b", "c", "d"]),
}
```

```python
df["one"]
```

```python
df["three"] = df["one"] * df["two"]
```

```python
pd.DataFrame(df)
```

Columns can be deleted or popped like with a dict:

```python
del df["two"]
```

```python
three = df.pop("three")
```

```python
pd.DataFrame(df)
```

```python
df["foo"] = "bar"
```

```python
pd.DataFrame(df)
```

When inserting a <u>`Series`</u> that does not to have the same index as the <u>`DataFrame`</u>, it will be conformed to the DataFrame's index:

```python
df["one_trunc"] = df["one"][:2]
```

```python
pd.DataFrame(df)
```

you can insert raw ndarrays but their length must match the length of the DataFrame's index.

By default, columns get inserted at the end. <u>`DataFrame.insert()`</u>inserts at a particular location in the columns:

```python
df.insert(1, "bar", df["one"])
```

```python
pd.DataFrame(df)
```
