## Name attribute

<u>`Series`</u> also has a `name` attribute:

```python
s = pd.Series(np.random.randn(5), name="something")
s
```

```python
s.name
```

The <u>`Series`</u> `name` can be assigned automatically in many cases, in particular when selectin a single column from <u>`DDataframe`</u>, the `name` will be assigned the column label.

You can rename a <u>`Series`</u> with the <u>`Pandas.Series.rename()`</u>method.

```python
s2 = s.rename("Different")
```

```python
s2.name
```

Note that `s` and `s2` refer to different objects.
