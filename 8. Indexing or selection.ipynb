## Indexing / selection

The basics of indexing are follows:

<table>
    <tr>
        <th>Operation</th>
        <th>Syntax</th>
        <th>Result</th>    
    </tr>
    <tr>
        <td>Select column</td>
        <td><code>
            df[col]
        </td>
        <td>Series</td>
    </tr>
    <tr>
        <td>Select row by label</td>
        <td><code>
            df.loc[label]
        </td>
        <td>Series</td>
    </tr>
    <tr>
        <td>Select row by integer location</td>
        <td><code>
            df.iloc[loc]
        </td>
        <td>Series</td>
    </tr>
    <tr>
        <td>Slice rows</td>
        <td><code>
            df[5:10]
        </td>
        <td>DataFrame</td>
    </tr>
    <tr>
        <td>Select rows by boolean vector</td>
        <td><code>
            df[bool_vec]
        </td>
        <td>DataFrame</td>
    </tr>
</table>
